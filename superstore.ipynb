```python
import os
import pandas as pd

file_path = r"C:\Users\Rethish Kumar\ROHPYTHON\Python class\Superstore project\Superstore_sample.csv"

df = pd.read_csv(file_path, encoding="latin1")

df.head()
```

```python
#DATA PREPARING

df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
df['Order Year']=df['Order Date'].dt.to_period('M')
```

```python
df['Shipping Days']=df['Ship Date']-df['Order Date']
```

```python
df.head()
```

```python
df.info()
```

```python
#•	Which category generates maximum sales? 

category_summary=df.groupby('Category')[['Sales','Profit']].sum().sort_values(by='Sales',ascending=False)
category_summary
```

```python
category_summary['Sales'].plot(kind='bar',figsize=(10,5))
plt.title("Sales by category")
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
category_summary['Profit'].plot(kind='bar',figsize=(10,5))
plt.title("Profit by category")
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
#Which regions perform best?
region_summary=df.groupby('Region')[['Sales','Profit']].sum().sort_values(by='Sales',ascending=False)
region_summary
```

```python
region_summary['Profit_%']=region_summary['Profit']/region_summary['Sales']*100
region_summary['Profit_%']
```

```python
region_summary['Profit'].plot(kind='bar',figsize=(10,5))
plt.title('Region wise profit')
plt.xlabel('Region')
plt.ylabel('Profit')
plt.xticks(rotation=0)
plt.show()
```

```python
region_summary['Sales'].plot(kind='bar',figsize=(10,5))
plt.title('Region wise sales')
plt.xlabel("Region")
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
#Chart for profit margin%
region_summary['Profit_%'].plot(kind='bar',figsize=(10,5))
plt.xlabel('Region')
plt.ylabel('Profit_%')
plt.xticks(rotation=0)
plt.show()
```

```python
#•	Customer segment contribution 
monthly_sales=df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum().reset_index()
monthly_sales
```

```python
plt.figure(figsize=(12,5))

plt.plot(monthly_sales['Order Date'].astype(str),
         monthly_sales['Sales'], marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
quaterly_sales=df.groupby(df['Order Date'].dt.to_period('Q'))['Sales'].sum().reset_index()
quaterly_sales
```

```python
plt.Figure(figsize=(10,6))
plt.plot(quaterly_sales['Order Date'].astype(str),quaterly_sales['Sales'],marker='o')
plt.title('Quaterly sales trend')
plt.xlabel('Quater')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
yearly_sales=df.groupby(df['Order Date'].dt.to_period('Y'))['Sales'].sum().reset_index()
yearly_sales
```

```python
plt.plot(yearly_sales['Order Date'].astype(str),yearly_sales['Sales'],marker='o')
plt.title('Yearly sales trend')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
plt.fill_between(
    quaterly_sales['Order Date'].astype(str),
    quaterly_sales['Sales'],
    alpha=0.4
)

plt.plot(
    quaterly_sales['Order Date'].astype(str),
    quaterly_sales['Sales'],
    marker='o'
)

plt.title('Monthly Sales Area Chart')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
#Find the most profitable Sub-Category
category_subcategory.sort_values(
    by='Profit',
    ascending=False
).head(10)
```
